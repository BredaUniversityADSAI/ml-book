In machine learning, it is a common convention to use **X** and **y** to represent the data:

* **X** (uppercase) refers to the **input features** — the independent variables used by the model to make predictions. It is typically a table or matrix where each row is a data sample and each column is a feature.
* **y** (lowercase) refers to the **target variable** — the value we want to predict. It can be categorical (for classification) or continuous (for regression).

This convention helps keep code consistent and readable across different datasets and models.

````python
# X is a pandas DataFrame containing the feature variables,
# and y is a pandas Series containing the target variable.
type(X), type(y)
````

This is the Breast Cancer Wisconsin dataset. In this dataset, **each row represents one patient**, and the data describe characteristics of a **breast tumor** found in that patient. The **input data (`X`)** contain **30 features**, which are numerical measurements describing the tumor’s shape, size, texture, and other properties obtained from medical images. The **target variable (`y`)** tells us whether the tumor is **benign (0)**, meaning not cancerous, or **malignant (1)**, meaning cancerous.

There are **569 patients in total**, with **357 benign** and **212 malignant** cases. The goal of a machine learning model is to learn from these examples so that, given the measurements of a new tumor, it can **predict whether it is likely to be benign or malignant**. Note that in machine learning, the term “positive class” (here, malignant = 1) simply means **the class being predicted**, not that it is “positive” in a good sense.

````python
# 357 tumors are labeled as malignant
# 212 tumors are labeled as benign
y.value_counts() # very useful to know this method, instead of writing a loop to count
````

Before building a machine learning model on a dataset, it’s important to start with a clear understanding of the data and the problem you are trying to solve. Ask yourself the following key questions:

* What is the problem? What do I need to predict, and based on which input features?
* Type of problem: Is it a classification (predicting categories) or a regression (predicting continuous values) task?
* If classification: How many classes are there, and what are their value counts?
* If regression: Min-max value of the target? (going forward you need to check the distribution)
* Data size: How many rows (samples) and columns (features) does the dataset contain?
* Feature types: Which features are categorical and which are numerical?
* Data quality: Are there any missing values that need to be handled?

Answering these questions helps you understand your dataset and choose appropriate preprocessing steps and models.

Based on our code above, we can answer these questions for the Breast Cancer Wisconsin dataset.

* What is the problem? The goal is to predict whether a breast tumor is benign (not cancerous) or malignant (cancerous) based on measurements of the tumor.
* Type of problem: This is a classification problem because we are predicting one of two categories.
* If classification: There are 2 classes — benign (0) and malignant (1).
* Value counts per class: There are 357 benign cases and 212 malignant cases.
* Data size: The dataset has 569 rows (patients) and 30 columns (numerical features).
* Feature types: All 30 features are numerical, describing properties such as radius, texture, smoothness, and symmetry of the tumor.
* Missing values: The dataset does not contain any missing values.
