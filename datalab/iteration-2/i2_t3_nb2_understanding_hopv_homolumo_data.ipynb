Note: In pandas, text data often shows up with the dtype **`object`**, because pandas is built on top of NumPy, which was originally designed for numbers, not text. NumPy didn’t have a special type for variable-length strings, so it stored them as generic Python objects, and pandas inherited this behavior. This means that when you see `object`, it usually means “a column of Python strings.” More recently, pandas introduced a dedicated **`string` dtype**, which makes working with text clearer and more consistent. In practice, both work, but using `string` is preferred in new projects since it avoids some problems with `object` columns and handles missing values better.

Now that we have parsed the text file, cleaned out empty values, and converted the columns to the correct data types, we can take a closer look at the dataset. While the data descriptor document specifies what each field represents, the meanings may not be immediately clear without a chemistry background. Below is a plain-language explanation of each column, prepared with the help of ChatGPT.

1. **Digital Object Identifier (DOI)**

   * A permanent identifier of the research paper where the experiment was published.
   * Useful for tracing back to the original experimental setup and conditions.

2. **InChIKEY of molecule**

   * A hashed version of the InChI (International Chemical Identifier) string.
   * Provides a unique, compact ID for the molecule.
   * Unlike SMILES, it is not used directly for feature extraction, but it ensures uniqueness and allows linking to chemical databases.

3. **Construction (Polymer / Molecule)**

   * Whether the active layer material is a single small molecule or a polymer.
   * This affects how molecules are processed experimentally, and may correlate with properties like stability or efficiency.

4. **Architecture**

   * The type of device structure used in the solar cell (e.g., bulk heterojunction).
   * Relevant for interpreting performance metrics, but not directly a molecular property.

5. **Complement**

   * The electron acceptor or donor material paired with the molecule in the solar cell.
   * For example, PCBM derivatives are often listed here.
   * This is important for device-level performance

6. **HOMO (Highest Occupied Molecular Orbital)**

   * The energy level of the highest filled electron orbital (in eV).
   * A molecular property that influences how easily the molecule donates electrons.
   * **Prediction target** for your ML task.

7. **LUMO (Lowest Unoccupied Molecular Orbital)**

   * The energy level of the lowest empty electron orbital (in eV).
   * Indicates how easily the molecule can accept electrons.
   * **Prediction target** for your ML task.

8. **Electrochemical gap**

   * The difference between HOMO and LUMO (HOMO − LUMO).
   * Can be measured experimentally, often with cyclic voltammetry.
   * Correlates with the open-circuit voltage of a device.

9. **Optical gap**

   * The band gap inferred from absorption spectra (UV-Vis).
   * Related but not identical to the HOMO-LUMO gap because of excitonic effects.

10. **PCE (Power Conversion Efficiency)**

    * Percentage of sunlight converted into electricity by the device.
    * A device-level metric, depends not only on the molecule but also the architecture and complement.

11. **VOC (Open-Circuit Voltage)**

    * The maximum voltage a device can deliver when no current flows.
    * Influenced by HOMO of donor and LUMO of acceptor.

12. **JSC (Short-Circuit Current Density)**

    * Current per unit area when the circuit is shorted.
    * Related to how well the molecule absorbs light and generates charge carriers.

13. **Fill Factor (FF)**

    * A measure of how “square” the current–voltage curve is.
    * Affects the efficiency of the solar cell.
