## 5. Evaluation

As you learned, in clustering, the evaluation phase is different from classification or regression because we do not have labels to check our results against. Instead, we evaluate how good the clusters are in two ways. First, we use measures such as the silhouette score or elbow method to see if the clusters are compact and well separated. Second, we look at the clusters from a business or domain perspective to check if they make sense and are useful for the original question. In other words, evaluation is about both the **quality of the clusters** and their **usefulness in practice**.

In this project, our goal is to use clustering as a supporting step for the prediction tasks (classification and regression). For instance, clustering might reveal a molecular property that varies across clusters, which can then be added as a new feature to improve predictive models. Another application is clustering data from different sources to examine whether the source itself has an influence on the clustering patterns.
