When we did a train/test split, the purpose was to evaluate our model on unseen data. However, once we are done with testing and want to actually use the model (for example, to submit predictions in a competition), we might want to retrain it on the entire dataset. Training on all the data could potentially improve performance, since the model sees more examples, but the downside is that we no longer have a test set to estimate its accuracy.

In the case of a Kaggle competition, the performance estimate can instead come from the public leaderboard score, which serves as feedback on how well your model is doing compared to others.
